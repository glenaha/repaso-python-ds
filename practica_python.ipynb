Tarea 1: Definición de Variables

- Define una variable para el nombre de un producto (string), su precio (float), la cantidad en stock (int) y si tiene descuento disponible (bool).
- Imprime una frase usando f-strings que resuma el producto.

```python
## Tarea 1: Definición de Variables

nombre_producto = "Yerba mate 1 kg"   # str (texto)
precio = 5099.99                      # float (número con decimales)
stock = 111                           # int (número entero)
tiene_descuento = True                # bool (verdadero/falso)

# Resumen del producto usando un f-string
print(f"Producto: {nombre_producto} | Precio: ${precio} | Stock: {stock} unidades | ¿Tiene descuento?: {tiene_descuento}")
```

Tarea 2: Lógica de Negocio (Bucle + Condicional)
- Crea una lista llamada precios con al menos 5 valores numéricos.
- Itera sobre la lista y, usando un condicional, imprime si cada precio es "Caro" (mayor a 100) o "Económico" (menor o igual a 100).

```python
## Tarea 2: Lógica de Negocio

# Lista con precios de distintos productos
precios = [55.0, 155.5, 99.99, 100, 320.25]

# Recorre cada precio de la lista y lo clasifica según su valor
for precio in precios:
    if precio > 100:
        print(f"{precio} -> Caro")
    else:
        print(f"{precio} -> Económico")
```

Tarea 3: Estructura de Datos Compleja
- Crea un diccionario llamado almacen donde las llaves sean nombres de productos y los valores sean su cantidad en stock.
- Agrega un nuevo producto al diccionario.
- Consulta el stock de un producto específico e imprímelo.

```python
## Tarea 3: Estructura de Datos Compleja

# Diccionario: cada producto (llave) tiene asociada su cantidad en stock (valor)
almacen = {
    "arroz": 50,
    "fideos": 80,
    "aceite": 25
}

# Agrega un producto nuevo al diccionario
almacen["azucar"] = 40

# Consulta el stock de un producto específico
print(f"Stock de fideos: {almacen['fideos']} unidades")

# Muestra el diccionario completo para verificar que se agregó el nuevo producto
print(almacen)
```

Tarea 4: Modularización con Funciones
- Escribe una función llamada resumen_estadistico que reciba un único parámetro: una lista de números. Es decir, def resumen_estadistico(numeros): y la llamás como resumen_estadistico([10, 20, 30]).
- La función debe devolver un diccionario que contenga: la suma total, el promedio y la cantidad de elementos.
- Llama a la función con una lista de ejemplo e imprime el resultado.

```python
## Tarea 4: Modularización con Funciones

# Función que recibe una lista de números y devuelve un resumen estadístico
def resumen_estadistico(numeros):
    cantidad = len(numeros)        # cuántos elementos tiene la lista
    suma_total = sum(numeros)      # suma de todos los elementos
    promedio = suma_total / cantidad
    # Devuelve los tres resultados juntos en un diccionario
    return {
        "suma_total": suma_total,
        "promedio": promedio,
        "cantidad": cantidad
    }

# Llama a la función con una lista de ejemplo e imprime el resultado
resultado = resumen_estadistico([10, 20, 30])
print(resultado)

# También se puede usar con la lista de la Tarea 2
print(resumen_estadistico(precios))
```
